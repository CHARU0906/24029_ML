### Exercise 1

Using K-Means clustering compress any image of size 396 x 396 x 3.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import image as mpimg
from sklearn.cluster import KMeans
```

```python
img = mpimg.imread('sample_image.jpg')
print(img.shape)

plt.imshow(img)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python
x = img.reshape(-1, 3)
print(x.shape)

print('unique colors before compression:', len(np.unique(x, axis=0)))
```

```python
k = 16
model = KMeans(n_clusters=k, init='k-means++', random_state=42)
y_means = model.fit_predict(x)

compressed = model.cluster_centers_[y_means]
compressed = compressed.reshape(img.shape).astype('uint8')

print('unique colors after compression:', len(np.unique(compressed.reshape(-1, 3), axis=0)))
```

```python
plt.imshow(compressed)
plt.title(f'Compressed Image (k={k})')
plt.axis('off')
plt.show()
```
