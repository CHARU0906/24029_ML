### Exercise 1

In a classification tree the dataset gets split based on its variables. Here we have two variables, age and income, which decide whether someone buys a house.

If the training data shows that 70% of people above age 30 bought a house, the tree will split on age first, so age becomes the first (root) node. This split makes the data about 80% pure, i.e mostly one class on either side of the split.

After the age split, the second node then looks at income to split the remaining data further.
